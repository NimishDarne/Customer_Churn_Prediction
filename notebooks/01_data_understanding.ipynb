Customer vs Churn

```python
# Create a contingency table
contingency_table = pd.crosstab(df['Contract'], df['Churn'], normalize='index')
print(contingency_table)
```

```python
from scipy.stats import chi2_contingency
# scipy's chi2_contingency function takes the contingency table 
# and computes the Chi-Square statistic and the p-value.
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\nChi-Square Statistic: {chi2:.4f}")
print(f"P-Value: {p_value:.4e}")
```

```python
if p_value < 0.05:
    print("Conclusion: Significant relationship exists! The two variables depend on each other.")
else:
    print("Conclusion: No significant relationship. The variations might be random noise.")
```

```python
# ROW-WISE PERCENTAGES (Normalize along rows)
# normalize='index' divides each cell count by the total count of that row.
# Multiply by 100 to convert decimals into percentage values.
row_percentages = pd.crosstab(df['Contract'],df['Churn'],normalize='index') * 100
print(row_percentages.round(2))
```

```python
plt.figure(figsize=(8, 5))

# sns.heatmap draws a color-coded matrix grid.
sns.heatmap(
    row_percentages,  # The percentage dataframe to plot
    annot=True,        # Display the percentage numbers inside each grid cell
    fmt=".1f",         # Format numbers to 1 decimal place (e.g., 75.4%)
    cmap="Blues",      # Use shades of blue (darker = higher value)
    cbar=True          # Show the color rating bar on the side
)

plt.title("Contract vc Churn")
plt.xlabel("Churn")
plt.ylabel("Contract")
plt.show()
```

```python
# .plot(kind='bar', stacked=True) converts a crosstab directly into a stacked bar plot
row_percentages.plot(
    kind='bar', 
    stacked=True, 
    figsize=(8, 5), 
    colormap='Set2'
)

plt.title("-")
plt.xlabel("Contract")
plt.ylabel("CHurn")
plt.legend(title="Contract vs Churn", bbox_to_anchor=(1.05, 1)) # Put legend outside plot
plt.xticks(rotation=0) # Keep category labels horizontal and easy to read
plt.show()
```
