```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['Churn'].value_counts()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 5))

# 3. Create the count plot in ONE line

ax = sns.countplot(data=df, x='Churn', hue='Churn', palette='viridis', legend=False)

# 4. Add the exact numbers on top of each bar automatically
ax.bar_label(ax.containers[0], padding=3)

# 5. Add titles and show
plt.title("Category Counts Summary", fontsize=14, pad=15)
plt.xlabel("Categories")
plt.ylabel("Count (Total: 7,043)")
plt.tight_layout()
plt.show()
```

```python
df['gender'].value_counts()
```

```python
sns.set_theme(style='whitegrid') # This specifically adds light gray background lines
plt.figure(figsize=(9,5))  #  This sets the physical dimensions of the window. The first number (9) is the width in inches, and the second (5) is the height in inches

'''sns.countplot(...): This tells Seaborn, "Look at this data, count the unique categories, and turn them into bars."data=df: This points Seaborn to your active DataFrame containing the 7,043 rows.
x='your_column_name': This tells Seaborn which column to look at. Putting it on x tells it to lay the categories side-by-side along the bottom (horizontal axis).
hue='your_column_name': This tells Seaborn to assign a distinct color to each unique category in that column.palette='viridis': A palette is a built-in color scheme. viridis is a highly popular option that shifts beautifully from deep purple to green and bright yellow.legend=False: Because our categories are already listed neatly on the bottom x-axis, a color legend box is redundant. Setting this to False hides the extra box and keeps the chart clean.'''
ay = sns.countplot(data=df, x='gender',hue='gender',palette='viridis',legend=False)

ay.bar_label(ay.containers[0],padding = 3)

'''ay.containers: This tells Python to look at the physical bars drawn on the chart.
ay.bar_label(...): This automatically calculates the height of each bar and writes the exact number directly on top of it.
padding=3: This leaves a tiny, 3-pixel vertical gap between the top edge of the bar and the text so they do not blend together.'''
plt.title("Gender Counts Summary", fontsize=14, pad=15)
plt.xlabel("gender")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
